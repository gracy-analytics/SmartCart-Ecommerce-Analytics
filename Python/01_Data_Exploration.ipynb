```python
# Create a working copy of Orders
orders_analysis = orders_clean.copy()

# Identify both date formats
slash_mask = orders_analysis["order_date"].str.contains("/", na=False)
hyphen_mask = orders_analysis["order_date"].str.contains("-", na=False)

# Create a new datetime column
orders_analysis["order_date_clean"] = pd.NaT

# Convert slash-format dates (M/D/YYYY)
orders_analysis.loc[slash_mask, "order_date_clean"] = pd.to_datetime(
    orders_analysis.loc[slash_mask, "order_date"],
    format="%m/%d/%Y"
)

# Convert hyphen-format dates (DD-MM-YYYY)
orders_analysis.loc[hyphen_mask, "order_date_clean"] = pd.to_datetime(
    orders_analysis.loc[hyphen_mask, "order_date"],
    format="%d-%m-%Y"
)

# Verify conversion
print(orders_analysis[["order_date", "order_date_clean"]].head(10))
print("\nData type:", orders_analysis["order_date_clean"].dtype)
print("Missing dates:", orders_analysis["order_date_clean"].isna().sum())
```

```python
# Inspect records with negative line profit

order_items.loc[
    order_items["line_profit_usd"] < 0,
    [
        "order_item_id",
        "quantity",
        "line_total_usd",
        "unit_cost_usd",
        "line_profit_usd"
    ]
].head(10)
```

```python
# Verify line profit calculations

order_items["calculated_profit"] = (
    order_items["line_total_usd"]
    - (order_items["unit_cost_usd"] * order_items["quantity"])
)

order_items["profit_difference"] = (
    order_items["line_profit_usd"]
    - order_items["calculated_profit"]
)

print("Profit calculation mismatches:",
      (order_items["profit_difference"].abs() > 0.02).sum())

print("\nSample negative-profit records:")
order_items.loc[
    order_items["line_profit_usd"] < 0,
    [
        "order_item_id",
        "quantity",
        "line_total_usd",
        "unit_cost_usd",
        "line_profit_usd",
        "calculated_profit"
    ]
].head(10)
```

```python
# Check numeric ranges in Returns

returns_numeric = returns_analysis.select_dtypes(include="number")

returns_numeric.agg(["min", "max"]).T.assign(
    negative_values=(returns_numeric < 0).sum()
)
```

```python
# Compare item quantities between Orders and Order Items

items_per_order = (
    order_items.groupby("order_id")["quantity"]
    .sum()
    .reset_index(name="calculated_items_count")
)

quantity_check = orders_clean[
    ["order_id", "items_count"]
].merge(
    items_per_order,
    on="order_id",
    how="left"
)

quantity_check["difference"] = (
    quantity_check["items_count"]
    - quantity_check["calculated_items_count"]
)

print("Total orders checked:", len(quantity_check))
print("Mismatched orders:", (quantity_check["difference"] != 0).sum())

quantity_check.loc[
    quantity_check["difference"] != 0
].head(10)
```

```python
# Verify order total calculations

orders_analysis["calculated_order_total"] = (
    orders_analysis["gross_amount_local"]
    - orders_analysis["discount_local"]
    + orders_analysis["shipping_fee_local"]
    + orders_analysis["tax_local"]
)

orders_analysis["total_difference"] = (
    orders_analysis["order_total_local"]
    - orders_analysis["calculated_order_total"]
)

print(
    "Order total mismatches:",
    (orders_analysis["total_difference"].abs() > 0.02).sum()
)
```

```python
# Verify local-to-USD order conversion

orders_analysis["calculated_order_total_usd"] = (
    orders_analysis["order_total_local"]
    / orders_analysis["fx_rate_to_usd"]
)

orders_analysis["usd_difference"] = (
    orders_analysis["order_total_usd"]
    - orders_analysis["calculated_order_total_usd"]
)

print(
    "USD conversion mismatches:",
    (orders_analysis["usd_difference"].abs() > 0.02).sum()
)
```

```python
# Verify returned quantities against purchased quantities

return_quantity_check = (
    returns_analysis.groupby("order_item_id")["quantity_returned"]
    .sum()
    .reset_index(name="total_quantity_returned")
    .merge(
        order_items[["order_item_id", "quantity"]],
        on="order_item_id",
        how="left"
    )
)

return_quantity_check["exceeds_purchased"] = (
    return_quantity_check["total_quantity_returned"]
    > return_quantity_check["quantity"]
)

print("Returned order items checked:", len(return_quantity_check))
print(
    "Items where returned quantity exceeds purchased quantity:",
    return_quantity_check["exceeds_purchased"].sum()
)
```

```python
# Bring the order exchange rate into the Returns dataset

returns_with_rate = returns_analysis.merge(
    orders_analysis[["order_id", "fx_rate_to_usd"]],
    on="order_id",
    how="left"
)

print("Total return records:", len(returns_with_rate))
print("Missing exchange rates:", returns_with_rate["fx_rate_to_usd"].isna().sum())
```

```python
# Verify local-to-USD refund calculations

returns_with_rate["calculated_refund_usd"] = (
    returns_with_rate["refund_amount_local"]
    / returns_with_rate["fx_rate_to_usd"]
)

returns_with_rate["refund_difference"] = (
    returns_with_rate["refund_amount_usd"]
    - returns_with_rate["calculated_refund_usd"]
)

print(
    "Refund conversion mismatches:",
    (returns_with_rate["refund_difference"].abs() > 0.02).sum()
)
```

```python
# Verify return duration calculations

return_date_check = returns_analysis.merge(
    orders_analysis[["order_id", "order_date_clean"]],
    on="order_id",
    how="left"
)

return_date_check["calculated_days"] = (
    return_date_check["return_date_clean"]
    - return_date_check["order_date_clean"]
).dt.days

return_date_check["date_difference"] = (
    return_date_check["days_to_return"]
    - return_date_check["calculated_days"]
)

print("Return records checked:", len(return_date_check))
print(
    "Date calculation mismatches:",
    (return_date_check["date_difference"] != 0).sum()
)
```

```python
# Inspect return date calculation mismatches

return_date_check.loc[
    return_date_check["date_difference"] != 0,
    [
        "return_id",
        "order_id",
        "return_date",
        "return_date_clean",
        "order_date_clean",
        "days_to_return",
        "calculated_days",
        "date_difference"
    ]
].head(15)
```

```python
# Compare original order dates with converted dates

return_date_check[
    return_date_check["date_difference"] != 0
][
    ["order_id", "return_date", "days_to_return"]
].head(10).merge(
    orders_clean[["order_id", "order_date"]],
    on="order_id",
    how="left"
)
```

```python
# Calculate the implied order date from return date and recorded duration

return_date_check["implied_order_date"] = (
    return_date_check["return_date_clean"]
    - pd.to_timedelta(return_date_check["days_to_return"], unit="D")
)

return_date_check[
    return_date_check["date_difference"] != 0
][
    [
        "order_id",
        "return_date",
        "days_to_return",
        "order_date_clean",
        "implied_order_date"
    ]
].head(15)
```

```python
# Correct mixed order date formats

orders_analysis["order_date_corrected"] = pd.NaT

slash_mask = orders_analysis["order_date"].str.contains("/", na=False)
hyphen_mask = orders_analysis["order_date"].str.contains("-", na=False)

orders_analysis.loc[slash_mask, "order_date_corrected"] = pd.to_datetime(
    orders_analysis.loc[slash_mask, "order_date"],
    format="%d/%m/%Y"
)

orders_analysis.loc[hyphen_mask, "order_date_corrected"] = pd.to_datetime(
    orders_analysis.loc[hyphen_mask, "order_date"],
    format="%d-%m-%Y"
)

print("Missing corrected dates:",
      orders_analysis["order_date_corrected"].isna().sum())

print(orders_analysis[
    ["order_date", "order_date_corrected"]
].head(10))
```

```python
# Recheck return duration using corrected order dates

return_date_check = returns_analysis.merge(
    orders_analysis[["order_id", "order_date_corrected"]],
    on="order_id",
    how="left"
)

return_date_check["calculated_days"] = (
    return_date_check["return_date_clean"]
    - return_date_check["order_date_corrected"]
).dt.days

return_date_check["date_difference"] = (
    return_date_check["days_to_return"]
    - return_date_check["calculated_days"]
)

print("Return records checked:", len(return_date_check))
print(
    "Date calculation mismatches:",
    (return_date_check["date_difference"] != 0).sum()
)
```
